```python
dialog = Desktop(backend="win32").window(title_re=".*Picture export.*")
dialog.wait("visible", timeout=10)
dialog.set_focus()

print(dialog.window_text())
```

```python
dialog.print_control_identifiers()
```

```python
buttons = []

for ctrl in dialog.descendants():
    try:
        if ctrl.friendly_class_name() == "Button":
            print(
                "Text:", ctrl.window_text(),
                "| Enabled:", ctrl.is_enabled(),
                "| Rect:", ctrl.rectangle()
            )
            buttons.append(ctrl)
    except Exception as e:
        print("Skipped control:", e)
```
